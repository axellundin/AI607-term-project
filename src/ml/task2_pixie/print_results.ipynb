```python
import matplotlib.pyplot as plt


alpha_values = [0.3, 0.4, 0.5, 0.6]
alpha_scores = [0.103911, 0.106536, 0.109354, 0.111801]

plt.figure()
plt.plot(alpha_values, alpha_scores, marker='o')
plt.xlabel(r'$\alpha$ (restart probability)')
plt.ylabel('Normalized DCG-weighted Recall@50')
plt.title('Effect of $\\alpha$ on User-level Recommendation')
plt.grid(True)
plt.tight_layout()
plt.savefig('ablation_alpha.pdf')
# plt.show()  # 로컬에서 확인하고 싶으면 사용
```
